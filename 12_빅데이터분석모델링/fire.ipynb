```python
import pandas as pd

df = pd.read_csv('소방청_화재발생 정보_20241231.csv', encoding='cp949')
print(df.shape)
df.head()
```

```python
# 인명피해(명)소계가 1 이상이면 1, 아니면 0 → 인명피해 발생 여부
df['인명피해_발생여부'] = (df['인명피해(명)소계'] > 0).astype(int)

df['인명피해_발생여부'].value_counts(normalize=True)  # 불균형 체크
```

```python
df['화재발생년원일'] = pd.to_datetime(df['화재발생년원일'])

df['발생_월'] = df['화재발생년원일'].dt.month        # 계절 확인용
df['발생_시간'] = df['화재발생년원일'].dt.hour           # 시간대
df['발생_요일'] = df['화재발생년원일'].dt.dayofweek   # 0=월요일, 6=일요일
```

```python
df.head()
```

```python
# 컬럼 정리
drop_cols = ['인명피해(명)소계', '사망', '부상', '재산피해소계', '부동산', '동산', '화재발생년원일', '장소중분류', '장소소분류', '시군구']
df = df.drop(columns=drop_cols)
```
