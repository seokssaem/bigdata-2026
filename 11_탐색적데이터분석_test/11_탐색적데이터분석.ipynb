```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# 그래프 한글 설정(윈도우)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 데이터 불러오기
df = pd.read_csv('prj_data_eda.csv')

df.head()
```

```python
# 데이터 구조확인
print("[데이터 크기] :", df.shape)

print("\n[데이터 타입]")
print(df.dtypes)

```

```python
print("\n [기술통계량]")
df.describe()
```

```python
# 수치형 변수
num_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()

# 범주형 변수(str + bool)
cat_cols = df.select_dtypes(include=['str','bool']).columns.tolist()

print("수치형 변수 :", num_cols)
print("범주형 변수 :", cat_cols)
```

```python
df['쿠폰사용여부_Y'].value_counts()
```

```python
# 수치형변수 '성별_남'의 히스토그램
plt.figure(figsize=(8,5))
sns.histplot(df['쿠폰사용여부_Y'], bins=30)
plt.title('쿠폰사용여부_Y 히스토그램')
plt.show()
```

```python
# 분포곡선이 있는 수치형변수 '나이'의 히스토그램
sns.histplot(df['쿠폰사용여부_Y'], bins=30, kde=True)
```

-결과 : 쿠폰 사용 고객과 미사용 고객의 비율을 확인할 수 있으며, 두 그룹의 분포가 크게 치우쳐 있는지 확인함.

```python
# 범주형 변수 '쿠폰사용여부_Y' Countplot
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='쿠폰사용여부_Y')
plt.show()
```

```python
# 수치형 변수 '월평균구매금액'의 히스토그램
plt.figure(figsize=(7,5))
sns.histplot(df['월평균구매금액'], bins=30, kde=True)
plt.title('월평균구매금액 분포')
plt.show()
```

```python
# 수치형 변수 '월평균구매금액'의 히스토그램
plt.figure(figsize=(7,5))
sns.histplot(df['구매횟수'], bins=20, kde=True)
plt.title('구매횟수 분포')
plt.show()
```

### 월평균구매금액과 구매횟수는 일부 고객이 높은 구매금액을 기록하여 오른쪽으로 긴 꼬리를 가진 분포를 보였다.

### 쿠폰 사용 여부는 두 그룹의 분포를 확인할 수 있었으며 특정 그룹에 과도하게 치우치지 않았다.

```python
# 상관계수
corr = df[num_cols].corr()

corr
```

```python
# 히트맵
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("상관관계 Heatmap")
plt.show()
```

```python
# 히트맵에서 가장 높은 상관관계를 가지는 (월평균구매금액 ↔ 구매횟수) 의 산포도
plt.figure(figsize=(6,5))
sns.scatterplot(data=df, x='구매횟수', y='월평균구매금액')
plt.title('구매횟수와 월평균구매금액')
plt.show()
```

### 구매횟수와 월평균구매금액은 매우 강한 양의 상관관계를 보였다.

### 그러나 상관관계가 높다고 해서 구매횟수가 구매금액의 직접적인 원인이라고 단정할 수는 없으며, 
### 두 변수 모두 고객의 소비성향 등의 다른 요인의 영향을 받을 가능성이 있다.

```python
# 쿠폰사용여부별 구매금액 비교 박스플롯
plt.figure(figsize=(6,5))
sns.boxplot(data=df,  x='쿠폰사용여부_Y', y='월평균구매금액')
plt.title('쿠폰 사용 여부에 따른 구매금액')
plt.show()
```

```python
# 쿠폰사용여부에 따라 groupby()

group = df.groupby('쿠폰사용여부_Y').agg({
    '월평균구매금액':'mean',
    '구매횟수':'mean',
    '평균구매단가':'mean',
    '리뷰점수':'mean'
})

print(group)
```

```python
group.plot(kind='bar', figsize=(8,5))

plt.title('쿠폰 사용 여부에 따른 평균 비교')
plt.ylabel('평균')
plt.xticks([0,1],['미사용','사용'],rotation=0)
plt.show()
```

### 최종 해석 : 그룹별 비교

- 쿠폰 사용 고객과 미사용 고객의 평균 구매 특성을 비교한 결과, 두 그룹 간 월평균구매금액과 구매횟수에서 차이가 나타났다.

### 인사이트 

- 쿠폰을 사용하는 고객은 구매 활동이 활발한 경향을 보였다. 따라서 쿠폰 제공 정책은 고객의 구매를 촉진하는 마케팅 전략으로 활용될 수 있다.

- 구매횟수와 월평균구매금액은 높은 양의 상관관계를 보였다. 따라서 반복 구매 고객을 대상으로 멤버십 혜택을 강화하면 매출 증가에 도움이 될 것으로 판단된다.
