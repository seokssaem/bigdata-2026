```python
id_cols = ['주차장명', '소재지지번주소', '주차구획수']
df_long = pd.melt(
    df_raw,
    id_vars=id_cols
)

df_long = df_long.reset_index(drop=True)
df_long.head()
```

```python
df_long = df_long.drop('variable',axis=1)
df_long = df_long.drop('value',axis=1)
```

```python
df_long
```

```python
# 규모 구분 컬럼 추가
df_series = df_long['주차구획수'].copy()
for i, col in enumerate(df_series):
    if col < 20:
        df_series[i] = 0
    elif col <= 50:
        df_series[i] = 1
    else:
        df_series[i] = 2

df_series
```

```python
df_long['규모구분'] = df_series.map({
    0:'소형',
    1:'중형',
    2:'대형'
})

df_long
```

```python
# 무료/유료 구분
df_raw['요금정보']
```

```python
df_long['fee_type'] = df_raw['요금정보']

df_long['fee_type'] = df_long['fee_type'].fillna('유료')

df_long
```

```python
# 동(洞) 이름 추출 컬럼 추가
df_series = df_long['소재지지번주소'].copy()

for i, col in enumerate(df_series):
    df_series[i]=col.split()[2]
df_series
```

```python
df_long['동이름'] = df_series

df_long
```

```python
# 선택한 데이터를 PostgreSQL parkingdb에 parking_lot이라는 이름의 테이블로 저장하시오.

# DB 연결
DB_URL = 'postgresql://postgres:1234@localhost:5432/parkingdb'
TABLE_NAME = 'parking_lot'

def save_to_postgresql(df, db_url, tabel_name):
    df_save = df.copy()

    for col in df_save.columns:
        if pd.api.types.is_string_dtype(df_save[col]):
            df_save[col] = df_save[col].astype(str)

    engine = create_engine(db_url)

    with engine.connect() as conn:
        version = conn.execute(text('SELECT version();')).fetchone()[0]
        print('연결')
        print(f'{version[:80]}')

    df_save.to_sql(
        name=tabel_name,
        con=engine,
        if_exists='replace',
        index=False,
        chunksize=1000,
        method='multi'
    )

    with engine.connect() as conn:
        saved_count = conn.execute(text(f'SELECT COUNT(*) FROM {tabel_name};')).fetchone()[0]

        print(saved_count)
```

```python
# 함수 호출
save_to_postgresql(df_long, DB_URL, TABLE_NAME)
```
