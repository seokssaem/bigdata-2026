```python
# 문제3_1
iris=pd.read_csv('iris.csv')
iris
iris.sort_values(by='SepalWidth',ascending=False)
```

```python
#문제3_2
import matplotlib.pyplot as plt

newdf=iris.sort_values(by='SepalWidth',ascending=False)
top10=newdf.head(10)
top10
top10.plot(kind='bar', rot=0)
plt.title('TOP10')
plt.xlabel('Data')
plt.show()
```

```python
titanic=pd.read_csv('train.csv')
titanic
```

```python
# 문제4_1
titanic[['Sex', 'Survived']].groupby(['Sex']).mean().sort_values(by='Survived', ascending=False)
```

```python
# 문제4_2
titanic[['Parch', 'Survived']].groupby(['Parch'], as_index=False).mean().sort_values(by='Survived', ascending=False)
```

```python
titanic.dtypes
```

```python
#문제4_3
# 데이터프레임.groupby(그룹이되는컬럼)[계산하고 싶은 컬럼].집계함수()
titanic.groupby(['Pclass','Sex'])[['Survived','Age']].mean()
```

```python
#문제4_4
import seaborn as sns
sns.countplot(x='Embarked',hue='Survived',data=titanic)
plt.show()
```
