```python
import pandas as pd
from sqlalchemy import create_engine, text
import os

BASE_DIR = os.getcwd()
BASE_DIR
```

```python
DATA_PATH = os.path.join(BASE_DIR, 'input', 'paking.csv')
DATA_PATH
```

```python
OUTPUT_PATH = os.path.join(BASE_DIR, 'output', 'paking_long.csv')
OUTPUT_PATH
```

```python
DB_URL = 'postgresql://postgres:1234@localhost:5432/pakingdb'
```

```python
df_raw = pd.read_csv(DATA_PATH, encoding='cp949')
df_raw.head()
```

```python
id_cols = ['주차장명', '소재지지번주소', '주차구획수']
df_paking = df_raw[id_cols].copy()
print(f'선택한 컬럼 수 : {len(df_paking)}개')
print(f'선택한 컬럼 목록 : {df_paking}')
```

```python
df_paking.head()
```

```python
def save_to_postgresql(df, db_url, table_name):
    df_save = df.copy()

    for col in df_save.columns:
        if pd.api.types.is_string_dtype(df_save[col]):
            df_save[col] = df_save[col].astype(str)
    engine = create_engine(db_url)

    with engine.connect() as conn:
        version = conn.execute(text('SELECT version();')).fetchone()[0]
        print('PostgreSQL 연결 성공!')
        print(version[:80] + '....')

    df_save.to_sql(
        name=table_name,
        con=engine,
        if_exists='replace',
        index=False,
        chunksize=1000,
        method='multi'
    )

    with engine.connect() as conn:
        saved_count = conn.execute(text(f'SELECT COUNT(*) FROM {table_name};')).fetchone()[0]

    print(f'저장 완료! {saved_count:,}행')
    print(f'DB: pakingdb / table : {table_name}')
```

```python
TABLE_NAME = 'paking_lot'
```

```python
save_to_postgresql(df_paking, DB_URL, TABLE_NAME)
```

```python
df_paking.to_csv(OUTPUT_PATH, encoding='utf-8-sig', index=False)

print('CSV 저장 완료!')
print(f'경로: {OUTPUT_PATH}')
print(f'크기: {len(df_paking):,}행 x {len(df_paking.columns)}열')
```
