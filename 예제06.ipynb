# 타이타닉 데이터

## 데이터 확인하기

```python
import pandas as pd
titanic = pd.read_csv('train.csv')
titanic.head()
```

```python
titanic.info()
```

```python
titanic.describe()
```

```python
titanic.sort_values('Fare',ascending=False) # 내림차순
```

```python
pd.DataFrame(titanic['Survived'].value_counts())
```

## 데이터 전처리

```python
titanic.isnull().sum()
```

```python
titanic.drop(['Cabin'],axis=1,inplace=True)
```

```python
titanic.columns
```

```python
titanic['Embarked'].value_counts() # 최빈값 'S'
```

```python
titanic['Embarked']=titanic['Embarked'].fillna('S')
titanic.isnull().sum()
```

```python
avg=titanic['Age'].mean()
titanic['Age']=titanic['Age'].fillna(avg)
titanic.isnull().sum()
```

```python
# 항구별('Embarked') Pclass 컬럼의 평균값 출력
titanic.groupby('Embarked')['Pclass'].mean()
```

```python
# ['Pclass', 'Sex'] 별 Age 컬럼의 평균값 출력
titanic.groupby(['Pclass', 'Sex'])['Age'].mean()
```
